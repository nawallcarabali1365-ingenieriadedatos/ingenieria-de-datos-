```python

# EDA — VENTAS TIENDITA

import pandas as pd
import matplotlib.pyplot as plt
```

```python

#  CARGAR DATOS tiendita
from google.colab import files

# Subir el archivo
subido = files.upload()

# Leer el archivo
nombre_archivo = list(subido.keys())[0]

df = pd.read_excel(nombre_archivo)

df
```

```python

# lectura de datos (tipos) ENTENDIMIENTO INICIAL

print("Número de filas y columnas:")
print(df.shape)

print("\nTipos de datos:")
print(df.dtypes)

print("\nPrimeras filas:")
display(df.head())
```

```python

# nulos y duplicados

print("Valores nulos por columna:")
display(df.isnull().sum())

print("\nNúmero de filas duplicadas:")
print(df.duplicated().sum())
```

```python
print("\nInformación general del dataset:")
df.info()
```

```python
print("\nInformación general del dataset:")
df.info()
```

```python

#  ANÁLISIS UNIVARIADO

df.describe()
```

```python
df[
    [
        "precio vendido",
        "precio real",
        "diferencia",
        "porcentaje de descuento"
    ]
].describe()
```

```python
print("Precio promedio vendido:",
      round(df["precio vendido"].mean(), 3))

print("Precio real promedio:",
      round(df["precio real"].mean(), 3))

print("Diferencia promedio:",
      round(df["diferencia"].mean(), 3))

print("Descuento promedio:",
      round(df["porcentaje de descuento"].mean() * 100, 2), "%")
```

```python
df["precio vendido"].plot(
    kind="hist",
    bins=8,
    title="Distribución del precio vendido"
)

plt.xlabel("Precio vendido")
plt.ylabel("Frecuencia")
plt.show()
```

```python
(df["porcentaje de descuento"] * 100).plot(
    kind="hist",
    bins=8,
    title="Distribución del porcentaje de descuento"
)

plt.xlabel("Descuento (%)")
plt.ylabel("Frecuencia")

plt.show()
```

```python

# ANÁLISIS BIVARIADO

precio_producto = (
    df.groupby("producto")["precio vendido"]
      .mean()
      .sort_values(ascending=False)
)

precio_producto
```

```python
precio_producto.plot(
    kind="bar",
    figsize=(12, 6),
    title="Precio promedio vendido por producto"
)

plt.xlabel("Producto")
plt.ylabel("Precio promedio vendido")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
descuento_producto = (
    df.groupby("producto")["porcentaje de descuento"]
      .mean()
      .sort_values(ascending=False)
)

descuento_producto
```

```python
(descuento_producto * 100).plot(
    kind="bar",
    figsize=(12, 6),
    title="Descuento promedio por producto"
)

plt.xlabel("Producto")
plt.ylabel("Descuento promedio (%)")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
df.plot.scatter(
    x="precio real",
    y="precio vendido",
    title="Precio real vs precio vendido"
)

plt.xlabel("Precio real")
plt.ylabel("Precio vendido")

plt.show()
```

```python

#CORRELACIÓN

correlacion = df[
    ["precio real", "precio vendido"]
].corr()

display(correlacion)
```

```python
r = df["precio real"].corr(df["precio vendido"])

print("Coeficiente de correlación:", round(r, 3))
```

```python
df.plot.scatter(
    x="precio real",
    y="precio vendido",
    figsize=(7, 5),
    title="Correlación entre precio real y precio vendido"
)

plt.xlabel("Precio real")
plt.ylabel("Precio vendido")

plt.show()
```

El coeficiente de correlación es aproximadamente 0,983, lo que indica una relación lineal positiva y muy fuerte entre el precio real y el precio vendido. En general, cuando aumenta el precio real de un producto, también aumenta su precio de venta.

# Conclusiones

### Hallazgo 1

El precio real y el precio vendido presentan una relación lineal positiva y muy fuerte, con un coeficiente de correlación de aproximadamente 0,983. Esto indica que los productos con precios reales más altos tienden a presentar también precios de venta más altos.

### Hallazgo 2

El precio promedio de venta fue de 9,835, mientras que el precio real promedio fue de 11,525. La diferencia promedio fue de 1,69 y el descuento promedio fue aproximadamente del 20,31 %. Esto muestra que, en promedio, los productos fueron vendidos por debajo de su precio real.

### Hipótesis a validar

Los productos con mayor precio real tienden a conservar un mayor precio de venta, incluso después de aplicar descuentos. Esta hipótesis puede validarse mediante el análisis de correlación entre el precio real y el precio vendido.
